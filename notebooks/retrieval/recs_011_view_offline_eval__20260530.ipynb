# View offline eval results (read-only)

Loads CSVs written by `recs_job_eval_retrieval.py`. **Does not re-score** examples.

Prerequisite: `python scripts/recs_job_eval_retrieval.py configs/recs_job_eval_retrieval.json` (optionally with `--examples-parquet`).

## Retrieval vs ranking (read this first)

Each method gets **two** metric tables from the same score vector:

| Stage | CSV prefix | k | Question | Primary metrics |
|-------|------------|---|----------|-----------------|
| **Retrieval** | `eval_retrieval_*` | `k_retrieval=100` | Are the right games in the candidate pool? | `Recall@K`, `Hit@K` |
| **Ranking** | `eval_ranking_*` | `k_final=10` | Is the top-10 **order** good? | Slice A: `NDCG@K`; Slice B: `Hit@K` |

**Oracle columns** (`OracleHit@K`, `OracleNDCG@K`) appear only on **ranking** tables. They show the best order possible within each method's top-100 pool. Large `(OracleNDCG@K − NDCG@K)` on `two_tower_v1` means **ranker headroom**, not a broken retriever.

Contract: [`docs/recommendation_evaluation_overview.md`](../../docs/recommendation_evaluation_overview.md)

```python
from __future__ import annotations

import json
from pathlib import Path

import pandas as pd
from IPython.display import display

# --- pick run folder ---
# "latest"  -> artifacts/recs/offline_eval/runs/latest
# or a named archive, e.g. "20260526_144828__baseline_retrieval"
EVAL_RUN = "latest"

REPO_ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").is_file())
RUNS_ROOT = REPO_ROOT / "artifacts/recs/offline_eval/runs"
EVAL_DIR = RUNS_ROOT / "latest" if EVAL_RUN == "latest" else RUNS_ROOT / EVAL_RUN

if not EVAL_DIR.is_dir():
    raise FileNotFoundError(f"Eval run dir not found: {EVAL_DIR}")

PATHS = {
    "ranking_overall": EVAL_DIR / "eval_ranking_overall.csv",
    "ranking_by_slice": EVAL_DIR / "eval_ranking_by_slice.csv",
    "retrieval_overall": EVAL_DIR / "eval_retrieval_overall.csv",
    "retrieval_by_slice": EVAL_DIR / "eval_retrieval_by_slice.csv",
    "run_meta": EVAL_DIR / "eval_offline_run_meta.json",
}
print(f"EVAL_DIR={EVAL_DIR}")
```

```python
def _load_csv(key: str) -> pd.DataFrame | None:
    path = PATHS[key]
    if not path.is_file():
        print(f"skip {key}: missing {path.name}")
        return None
    return pd.read_csv(path)

rank_overall = _load_csv("ranking_overall")
rank_slice = _load_csv("ranking_by_slice")
retr_overall = _load_csv("retrieval_overall")
retr_slice = _load_csv("retrieval_by_slice")

run_meta = None
if PATHS["run_meta"].is_file():
    run_meta = json.loads(PATHS["run_meta"].read_text(encoding="utf-8"))
    print("methods:", run_meta.get("methods_requested"))
    print(
        "k:",
        f"k_final={run_meta.get('k_final')}",
        f"k_retrieval={run_meta.get('k_retrieval')}",
    )
else:
    print("no eval_offline_run_meta.json")
```

```python
from IPython.display import Markdown, display

# --- RANKING @ k_final (order-sensitive; oracle ceiling columns here) ---
display(Markdown("### Ranking overall (`eval_ranking_overall.csv`) — top-**10** order"))
if rank_overall is not None:
    sort_cols = [c for c in ["NDCG@K", "MAP@K", "MRR", "Hit@K"] if c in rank_overall.columns]
    display(rank_overall.sort_values(sort_cols, ascending=False))
else:
    print("ranking overall: missing")

# --- RETRIEVAL @ k_retrieval (pool membership; no oracle columns) ---
display(Markdown("### Retrieval overall (`eval_retrieval_overall.csv`) — top-**100** pool"))
if retr_overall is not None:
    sort_cols = [c for c in ["Recall@K", "Precision@K", "Hit@K"] if c in retr_overall.columns]
    display(retr_overall.sort_values(sort_cols, ascending=False))
else:
    print("retrieval overall: missing")
```

```python
display(Markdown("### Ranking by slice (`eval_ranking_by_slice.csv`) — Slice A primary: `NDCG@K`"))
if rank_slice is not None:
    display(rank_slice.sort_values(["slice_name", "NDCG@K"], ascending=[True, False]))
else:
    print("ranking by slice: missing")

display(Markdown("### Retrieval by slice (`eval_retrieval_by_slice.csv`) — primary: `Recall@K`"))
if retr_slice is not None:
    display(retr_slice.sort_values(["slice_name", "Recall@K"], ascending=[True, False]))
else:
    print("retrieval by slice: missing")
```

```python
try:
    import matplotlib.pyplot as plt
except ImportError:
    plt = None

if plt is None:
    print("matplotlib not available; tables above are enough.")
else:
    if rank_overall is not None and "NDCG@K" in rank_overall.columns:
        df = rank_overall.sort_values("NDCG@K", ascending=True)
        fig, ax = plt.subplots(figsize=(8, max(3, 0.4 * len(df))))
        ax.barh(df["method"], df["NDCG@K"], color="steelblue")
        ax.set_xlabel("NDCG@K @ k_final=10")
        ax.set_title(f"RANKING overall — {EVAL_RUN}")
        plt.tight_layout()
        plt.show()

    if retr_overall is not None and "Recall@K" in retr_overall.columns:
        df = retr_overall.sort_values("Recall@K", ascending=True)
        fig, ax = plt.subplots(figsize=(8, max(3, 0.4 * len(df))))
        ax.barh(df["method"], df["Recall@K"], color="darkorange")
        ax.set_xlabel("Recall@K @ k_retrieval=100")
        ax.set_title(f"RETRIEVAL overall — {EVAL_RUN}")
        plt.tight_layout()
        plt.show()

    # Ranker headroom: two_tower_v1 only (ranking stage)
    if rank_slice is not None and {"slice_name", "method", "NDCG@K", "OracleNDCG@K"}.issubset(rank_slice.columns):
        tt_a = rank_slice.query("slice_name == 'slice_a_multi_target' and method == 'two_tower_v1'")
        if not tt_a.empty:
            ndcg = float(tt_a["NDCG@K"].iloc[0])
            oracle = float(tt_a["OracleNDCG@K"].iloc[0])
            display(Markdown(
                f"**Ranker headroom (Slice A, `two_tower_v1`):** "
                f"NDCG@10={ndcg:.4f}, OracleNDCG@10={oracle:.4f}, gap={oracle - ndcg:.4f}"
            ))
```
